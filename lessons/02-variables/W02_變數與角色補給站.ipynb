## 3–11 分｜① 建立與使用變數

### 幫資料取個名字

`player_name` 代表角色名字，`coins` 代表金幣。這些名稱叫做**變數**，讓程式能透過名字找到資料。

看 `coins = 100` 時，從右邊往左讀：「把值 100 指定給名稱 coins」。`=` 是**指定值**，不是數學的相等判斷。

先看下一格，猜它會顯示哪個角色、有幾枚金幣，再執行。


````python
player_name = "星際飯糰"
coins = 100
print("角色：", player_name)
print("金幣：", coins)
````

預期顯示「角色： 星際飯糰」與「金幣： 100」。

- `"星際飯糰"` 是文字，也叫**字串**，外面要有引號；`100` 是**整數**，可以直接計算。金幣請寫 `100`，不要加引號變成文字 `"100"`。
- `print(coins)` 會取用變數的值；`print("coins")` 則只會顯示 coins 這幾個字。
- `print("角色：", player_name)` 用逗號把標籤與資料放在同一行顯示，中間會有空格；不需要把數字接成文字。

先建立變數，再使用它。這裡先看懂示範，自己的角色資料會在最後的同一張作品卡修改。

## 16–23 分｜③ 更新資料，讓金幣會計算

同一個名字可以重新指定新的值。例如 `coins = 120`，就改成以 coins 取用 120。

也可以先算再指定：`coins = coins + 20` 是先讀取右邊的舊金幣、加上 20，再把結果指定回 coins。

下一格讓角色領到 20 枚獎勵，買兩份各 30 枚的補給，再看看餘額平分給兩人的數量。**先口算餘額，再執行一次確認。**

| 符號 | 意思 | 本格用途 |
| --- | --- | --- |
| `+` | 加 | 金幣加上獎勵 |
| `*` | 乘 | 單價乘以兩份 |
| `-` | 減 | 金幣扣掉花費 |
| `/` | 除 | 餘額平分給兩人 |


````python
coins = 100
coins = coins + 20
cost = 30 * 2
remaining = coins - cost
print("領獎後金幣：", coins)
print("兩份補給花費：", cost)
print("購買後餘額：", remaining)
print("餘額平分兩人：", remaining / 2)
````

結果依序是 **120、60、60、30.0**。`/` 會產生小數型別的結果，所以整除時也會看見 `30.0`。

這格開頭會重新設定 100 枚金幣，重跑整格仍會得到同樣結果。修改程式後要再執行；電腦不會因為你改了文字，就自動重算已經顯示的輸出。

## 23–28 分｜④ 常數：提醒大家「這個值預計固定」

補給站這次規定每份補給 30 枚金幣。我們用 **`UNIT_PRICE`** 表示單價：常數慣例使用全大寫，單字之間用底線。

**Python 不會因為全大寫就禁止修改。** 常數是給讀程式的人看的約定，不是另一種不能修改的資料型別。

下一格把固定單價與購買數量分開。先找出哪個名稱代表固定價格，再執行；這不是新作品。

## 28–38 分｜⑤ 交換裝備，要先保留原本的值

角色 A 的 `item_a` 是「地圖」，角色 B 的 `item_b` 是「手電筒」。交換後應該是 A 拿手電筒、B 拿地圖。

### 先看三步交換的原理

先用暫存名稱 `temp` 記住 A 原本的裝備，再交換：

```python
temp = item_a
item_a = item_b
item_b = temp
```

| 步驟 | item_a | item_b | temp |
| --- | --- | --- | --- |
| 原本 | 地圖 | 手電筒 | 尚未建立 |
| 先保留 A 的原值 | 地圖 | 手電筒 | 地圖 |
| A 改拿 B 的裝備 | 手電筒 | 手電筒 | 地圖 |
| B 取回保留的原值 | 手電筒 | 地圖 | 地圖 |

如果先做 `item_a = item_b`，再做 `item_b = item_a`，兩個名稱最後都會代表手電筒，因為 A 已經不是原本的地圖。

### 再看 Python 的簡潔寫法

`item_a, item_b = item_b, item_a` 會**先取得右邊兩個舊值，再依序指定給左邊**。左邊 A 得到原本的 B，左邊 B 得到原本的 A。

上面的三步只用來理解；執行下一格這一次交換就好，不用再抄另一份。


````python
item_a = "地圖"
item_b = "手電筒"
print("交換前，A：", item_a, "B：", item_b)
item_a, item_b = item_b, item_a
print("交換後，A：", item_a, "B：", item_b)
````

預設資料的餘額是 40，交換後 A 拿手電筒、B 拿地圖。改成自己的金幣與裝備後，依自己的資料核對。

## 48–50 分｜結論
